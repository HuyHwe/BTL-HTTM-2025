```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Load dataset MNIST
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
# Chuẩn hóa dữ liệu về [0,1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# One-hot encode labels
num_classes = 10
y_train = keras.utils.to_categorical(y_train, num_classes)
y_test = keras.utils.to_categorical(y_test, num_classes)
```

```python
# 2. Dense Neural Network (Fully Connected)
# -------------------------------
x_train_dense = x_train.reshape(-1, 28*28)  # Flatten ảnh 28x28 -> 784
x_test_dense = x_test.reshape(-1, 28*28)

dense_model = keras.Sequential([
    layers.Dense(512, activation="relu", input_shape=(784,)),
    layers.Dense(256, activation="relu"),
    layers.Dense(num_classes, activation="softmax")
])

dense_model.compile(optimizer="adam",
                    loss="categorical_crossentropy",
                    metrics=["accuracy"])

print("Training Dense NN...")
history_dense = dense_model.fit(x_train_dense, y_train,
                                validation_split=0.1,
                                epochs=10,
                                batch_size=128,
                                verbose=2)
```

```python
# 3. Convolutional Neural Network (CNN)
# -------------------------------
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)

cnn_model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation="relu"),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(num_classes, activation="softmax")
])

cnn_model.compile(optimizer="adam",
                  loss="categorical_crossentropy",
                  metrics=["accuracy"])

print("Training CNN...")
history_cnn = cnn_model.fit(x_train_cnn, y_train,
                            validation_split=0.1,
                            epochs=10,
                            batch_size=128,
                            verbose=2)
```

```python
# 3. Convolutional Neural Network (CNN)
# -------------------------------
from tensorflow.keras.optimizers import Adam
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)

cnn_model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation="relu"),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(num_classes, activation="softmax")
])



cnn_model.compile(
    optimizer=Adam(learning_rate=0.00001),   # <-- thay đổi learning rate tại đây
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


print("Training CNN...")
history_cnn = cnn_model.fit(x_train_cnn, y_train,
                            validation_split=0.1,
                            epochs=20,
                            batch_size=128,
                            verbose=2)
```

```python
# 4. Tính các chỉ số trên tập validation (MAE, MSE, RMSE)
# -------------------------------
def compute_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    return mae, mse, rmse

# Dự đoán trên tập validation
x_val_dense = x_train_dense[int(0.9*len(x_train)):]
y_val_dense = y_train[int(0.9*len(y_train)):]
dense_val_pred = dense_model.predict(x_val_dense)

x_val_cnn = x_train_cnn[int(0.9*len(x_train)):]
y_val_cnn = y_train[int(0.9*len(y_train)):]
cnn_val_pred = cnn_model.predict(x_val_cnn)

dense_mae, dense_mse, dense_rmse = compute_metrics(y_val_dense, dense_val_pred)
cnn_mae, cnn_mse, cnn_rmse = compute_metrics(y_val_cnn, cnn_val_pred)

print("\n===== Kết quả trên tập Validation =====")
print(f"Dense NN - MAE: {dense_mae:.4f}, MSE: {dense_mse:.4f}, RMSE: {dense_rmse:.4f}")
print(f"CNN      - MAE: {cnn_mae:.4f}, MSE: {cnn_mse:.4f}, RMSE: {cnn_rmse:.4f}")
```

```python
# 5. Vẽ biểu đồ Accuracy, Loss, MAE, MSE, RMSE
# -------------------------------
plt.figure(figsize=(15,10))

# Accuracy
plt.subplot(2,2,1)
plt.plot(history_dense.history["accuracy"], label="Dense train acc")
plt.plot(history_dense.history["val_accuracy"], label="Dense val acc")
plt.plot(history_cnn.history["accuracy"], label="CNN train acc")
plt.plot(history_cnn.history["val_accuracy"], label="CNN val acc")
plt.title("Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()

# Loss
plt.subplot(2,2,2)
plt.plot(history_dense.history["loss"], label="Dense train loss")
plt.plot(history_dense.history["val_loss"], label="Dense val loss")
plt.plot(history_cnn.history["loss"], label="CNN train loss")
plt.plot(history_cnn.history["val_loss"], label="CNN val loss")
plt.title("Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

# MAE, MSE, RMSE (bar chart so sánh sau train xong)
metrics_names = ["MAE", "MSE", "RMSE"]
dense_values = [dense_mae, dense_mse, dense_rmse]
cnn_values = [cnn_mae, cnn_mse, cnn_rmse]

x = np.arange(len(metrics_names))
width = 0.35

plt.subplot(2,2,3)
plt.bar(x - width/2, dense_values, width, label="Dense NN")
plt.bar(x + width/2, cnn_values, width, label="CNN")
plt.xticks(x, metrics_names)
plt.title("So sánh MAE, MSE, RMSE (Validation)")
plt.ylabel("Giá trị")
plt.legend()

plt.tight_layout()
plt.show()
```
